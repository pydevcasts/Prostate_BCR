```python
import numpy as np

arr = np.array([1,2,3,4,5])
print(arr)
print(f"Type: {type(arr)}")
print(f"Shape: {arr.shape}") #attr
print(f"Size: {arr.size}") #attr
print(f"Dimention: {arr.ndim}") #attr
print(dir(arr))
```

```python
import time

py_list = list(range(1000000)) #range builtin datatype 
np_array = np.arange(1000000)

start = time.time()
sum_py = sum(py_list)
end = time.time()
print(f"Python list sum:{sum_py}, Time: {end - start}")

start = time.time()
sum_np = np.sum(np_array)
end = time.time()
print(f"Numpy array sum:{sum_np}, Time: {end - start}")
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

concat = np.concatenate((arr1, arr2))
vstack = np.vstack((arr1, arr2))
hstack = np.hstack((arr1, arr2))

print("Concatenate:", concat)
print("Vertical Stack:\n", vstack)
print("Horizontal Stack:", hstack)
```
