```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 1. دو داده ساده تعریف می‌کنیم
data = {
    'قد': [170, 165, 180, 175, 160],
    'وزن': [70, 65, 80, 75, 60]
}

# 2. تبدیل به DataFrame
df = pd.DataFrame(data)

# 3. محاسبه ماتریس همبستگی (2x2)
corr_matrix = df.corr()
print("ماتریس همبستگی:")
print(corr_matrix)

# 4. رسم Heatmap ساده
plt.figure(figsize=(4, 3))
plt.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)

# تنظیمات ظاهری
plt.xticks(range(len(corr_matrix)), corr_matrix.columns, rotation=45)
plt.yticks(range(len(corr_matrix)), corr_matrix.columns)

# نمایش مقادیر روی خانه‌ها
for i in range(2):
    for j in range(2):
        plt.text(j, i, f'{corr_matrix.iloc[i, j]:.2f}', 
                 ha='center', va='center', color='white', fontsize=12)

plt.colorbar(label='همبستگی')
plt.title('Heatmap همبستگی بین قد و وزن')
plt.tight_layout()
plt.show()
```
