```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# داده‌های نمونه (با چند نویز)
data = [1, 2, 2, 3, 3, 4, 5, 6, 7, 8, 9, 10, 50]  # 50 نویز است

# تبدیل به سری pandas
s = pd.Series(data)

# محاسبهٔ چارک‌ها و IQR
Q1 = s.quantile(0.25)
Q3 = s.quantile(0.75)
IQR = Q3 - Q1

# محاسبهٔ مرزها
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# فیلتر کردن داده‌ها (حذف نویزها)
filtered_data = s[(s >= lower_bound) & (s <= upper_bound)]

# رسم دو نمودار برای مقایسه
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

# boxplot اصلی (با نویز)
ax[0].boxplot(s,vert=False)
ax[0].set_title('Boxplot with outlier')

# boxplot بدون نویز
ax[1].boxplot(filtered_data, vert=False)
ax[1].set_title('Boxplot withot outlier .delet 50')

plt.tight_layout()
plt.show()

print("داده‌های اصلی:", list(s))
print("داده‌های فیلترشده:", list(filtered_data))
```
