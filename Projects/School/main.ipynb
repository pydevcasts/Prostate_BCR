```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Data
data = {
    "Name": ["Ali", "Nastaran", "Amir", "Sara", "Reza", "Mina", "Hossein", "Taraneh"],
    "Math": [18, 13, 9, 15, 20, 14, 12, 16],
    "Science": [17, 19, 11, 13, 18, 15, 14, 12],
    "Language": [15, 14, 10, 16, 19, 12, 13, 17],
    "City": ["Tehran", "Shiraz", "Tabriz", "Mashhad", "Tehran", "Isfahan", "Tabriz", "Shiraz"]
}

df = pd.DataFrame(data)
```

```python

# -----------------------------------------------------------
# 🔹 Section 1 — Basic Data Inspection (Pandas)
# -----------------------------------------------------------

# 1️⃣ Print the dataframe
print("1️⃣ DataFrame:\n", df)


# 2️⃣ Print number of rows and columns
print("\n2️⃣ Shape of DataFrame:", df.shape)

# 3️⃣ Display data types of each column
print("\n3️⃣ Data types:")
print(df.info())

# 4️⃣ Print mean of each score column
print("\n4️⃣ Mean of each subject:")
print(df[["Math", "Science", "Language"]].mean())
```

```python

# -----------------------------------------------------------
# 🔹 Section 2 — Numerical Calculations (NumPy)
# -----------------------------------------------------------

# 5️⃣ Mean of Math scores using NumPy
math_mean = np.mean(df["Math"])
print("\n5️⃣ Mean of Math (NumPy):", math_mean)

# 6️⃣ Standard deviation of Language scores using NumPy
lang_std = np.std(df["Language"])
print("6️⃣ Standard deviation of Language (NumPy):", lang_std)

# 7️⃣ Total score of each student
total_scores = df[["Math", "Science", "Language"]].to_numpy().sum(axis=1)
print("7️⃣ Total scores of each student:")
print(total_scores)
```

```python


# -----------------------------------------------------------
# 🔹 Section 3 — Data Analysis (Pandas)
# -----------------------------------------------------------

# 8️⃣ Add a new column 'Average' showing mean of three subjects
df["Average"] = df[["Math", "Science", "Language"]].mean(axis=1)
print("\n8️⃣ DataFrame with Average column:\n", df)

# 9️⃣ Select students with average score above 15
top_students = df[df["Average"] > 15]
print("\n9️⃣ Students with average above 15:\n", top_students)

# 🔟 Count of students in each city
print("\n10️⃣ Number of students per city:")
print(df["City"].value_counts())

# 11️⃣ Find subject with highest average score
subject_means = df[["Math", "Science", "Language"]].mean()
top_subject = subject_means.idxmax()
print("\n11️⃣ Subject with highest average score:", top_subject)

# 2️⃣Find the lesson which has maximum score
print(f"\n2️⃣which lesson has maximum score", df.loc[df["Average"].idxmax(), ["Math", "Language", "Science"]].idxmax())
```

```python

# -----------------------------------------------------------
# 🔹 Section 4 — Data Visualization (Matplotlib)
# -----------------------------------------------------------

# 12️⃣ Bar chart of average scores per subject
plt.figure(figsize=(6,4))
plt.bar(subject_means.index, subject_means.values, color=['skyblue', 'lightgreen', 'salmon'])
plt.title("Average Scores by Subject")
plt.ylabel("Average Score")
plt.show()

# 13️⃣ Pie chart of student count per city
plt.figure(figsize=(5,5))
df["City"].value_counts().plot(kind="pie", autopct='%1.1f%%', startangle=120, colors=['gold', 'lightblue', 'lightcoral', 'plum'])
plt.title("Distribution of Students by City")
plt.ylabel("")  # remove y-label
plt.show()

# 14️⃣ Line chart of Math scores per student
plt.figure(figsize=(7,4))
plt.plot(df["Name"], df["Math"], marker='o', linestyle='-', color='purple')
plt.title("Math Scores by Student")
plt.xlabel("Student Name")
plt.ylabel("Math Score")
plt.grid(True)
plt.show()
```

```python
# -----------------------------------------------------------
# 🔹 Section 5 — Filter and Data Visualization (Matplotlib)
# -----------------------------------------------------------
# Filter students from Tehran with average > 16
filtered_students = df[(df["City"] == "Tehran") & (df["Average"] > 16)]
print("Filtered Students:\n", filtered_students)

# Plot the bar chart of their scores
plt.figure(figsize=(7,4))

# Extract subject scores
subjects = ["Math", "Science", "Language"]

for index, row in filtered_students.iterrows():
    plt.bar(subjects, row[subjects], label=row["Name"], alpha=0.7)

plt.title("Scores of Tehran Students with Average > 16")
plt.ylabel("Score")
plt.legend()
plt.show()
```
